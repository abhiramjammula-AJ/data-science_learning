```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB, GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_validate, ShuffleSplit
```

```python
df = pd.read_csv('D:\\Data science\\codes\\files\\mushroom.csv')
df.columns
```

```python
LE = LabelEncoder()
for i in df.columns:
    df[i] = LE.fit_transform(df[i])

df.head()
```

```python
# alternative
df = df.apply(LE.fit_transform)
df
```

```python
# X = df.iloc[:, 1:]
X = df.drop('Typeofmushroom', axis=1)
Y = df['Typeofmushroom']
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=100)
```

### Model Fit
naive baies

```python
model = MultinomialNB()
model.fit(X_train, Y_train)
Y_pred_train = model.predict(X_train)
Y_pred_test = model.predict(X_test)
```

```python
print("Train score =", accuracy_score(Y_train, Y_pred_train))
print("Test score =", accuracy_score(Y_test, Y_pred_test))
```

```python
SS = ShuffleSplit(n_splits=100, test_size=0.3, random_state=0)
cv_results = cross_validate(model, X, Y, cv=SS, scoring='accuracy', return_train_score=True)
print("accuracy of Train =",cv_results['train_score'].mean())
print('accuracy of test =', cv_results['test_score'].mean())
```
