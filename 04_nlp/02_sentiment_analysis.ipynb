```python
with open(file_path,'r',encoding='utf-8') as file:
    text_data = file.read()

print(text_data)
```

```python
import pandas as pd
import numpy as np
lines = text_data.split('\n')
df = pd.DataFrame({"text": lines})
print(df)
df.shape
```

```python
# remove the empty spaces
## check for null values -- 1
df = df[df['text'].notnull()]
df.shape
## check for no empty lines -- 2 --- all odd indexes have the data
odd_index_data = df.iloc[1::2]
print(odd_index_data)
```

```python
odd_index_data.reset_index(drop=True, inplace=True)
print(odd_index_data)
```

```python
df_clean = odd_index_data.loc[0:39]
df_clean.shape
```

```python
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()
```

```python
# .apply will apply the function to each row of the dataframe
df_clean['compound'] = df_clean['text'].apply(lambda x: analyzer.polarity_scores(x)['compound'])
df_clean.head()
```

```python
df_clean['sentiment'] = df_clean['compound'].apply(lambda x: 'positive' if x>0 else 'negative')
df_clean.head()
```

```python
df_clean['sentiment'].value_counts()
# target variable
```

```python
# Data preprocessing steps to clean the data before we can
# develop a model 
df_clean['text'] = df_clean['text'].str.lower() # normalization

import string
# remove punctuation
df_clean['text'] = df_clean['text'].str.replace('[{}]'.format(string.punctuation), '')

# lemmatization
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
df_clean['text'] = df_clean['text'].apply(lambda x: ' '.join([lemmatizer.lemmatize(word) for word in x.split()]))

# remove stopwords
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
df_clean['text'] = df_clean['text'].apply(lambda x: ' '.join([word for word in x.split() if word not in stop_words]))

df_clean.head()
```

```python
Y = df_clean['sentiment']
X = df_clean['text']
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(X)
X_new = X.toarray()
```

```python
X_new.shape
```

```python
from sklearn.naive_bayes import MultinomialNB
model = MultinomialNB()
model.fit(X_new, Y)
```

```python
y_pred = model.predict(X_new)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(Y, y_pred)
print("accuacy score: ",np.round(accuracy,2))
```
