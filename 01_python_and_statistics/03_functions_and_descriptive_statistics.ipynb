# Types of In-built Functions and Python Structure

- A file containing many functions is called a **module**.
- Many modules together form a **library**.
- Many libraries together form a **package**.
- Many packages together form a **framework**.

```mermaid

    Module --> Library;
    Library --> Package;
    Package --> Framework;
```

# Important Libraries for ML and AI in Data Science

## For Structured Data
- **NumPy**: Numerical Python for arrays (1D, 2D, 3D, ...) and all statistical operations.
- **Pandas**: Import and manipulate CSV files in Jupyter; data cleaning and manipulation.
- **Matplotlib & Seaborn**: Data visualization (box, bar, scatter plots, etc.).
- **SciPy, Statsmodels**: Probability, normal distribution, hypothesis testing, time series analysis.
- **Scikit-learn (`sklearn`)**: Machine Learning algorithms and utilities.
- **Streamlit, Flask**: Model deployment and building web apps.

## For Unstructured Data
- **NLTK (Natural Language Toolkit), spaCy**: Natural Language Processing (NLP).
- **TensorFlow, PyTorch, Keras**: Deep learning frameworks for image and text data.

### Importing Numpy

````python
import numpy as np
````

````python
age = np.array([24, 28, 28, 20, 26, 30])
age
````

````python
age = np.array([[24], [28], [28], [20], [26], [30]]) # Reshaping the array to a 2D array with one column 
# This allows for operations that require a 2D array/column vector to get in a correct form for statistical operations 
age
````

````python
data = np.array([[24, 171, 75, 100], [28, 175, 80, 110], [28, 180, 85, 120], [20, 165, 70, 90], [26, 170, 75, 95]]) # Example of a 2D array with multiple columns
data
````

````python
data.shape
````

````python
# access the columns
#          Row      ,   column
#      start: end-1 ,start: end-1
# data[     :       ,     :      ]

data[: , 1: 2]
````

````python
print("Mean: ", np.mean(age))
print("Median: ", np.median(age))
print("Standard Deviation: ", np.std(age))
print("Variance: ", np.var(age))
````

````python
print("Mean: ", np.mean(data[: , 0:1]))
print("Median: ", np.median(data[: , 0:1]))
print("Standard Deviation: ", np.std(data[: , 0:1]))
print("Variance: ", np.var(data[: , 0:1]))
print("min: ", np.min(data[: , 0:1]))
print("max: ", np.max(data[: , 0:1]))
````

````python
np.round(np.std(data[: , 0:1]), 2)  # Example of rounding the standard deviation to 2 decimal places
````

````python
np.sqrt(np.var(data[: , 0:1]))  # Example of calculating the standard deviation from variance
````

````python
print(np.sqrt(81)) 
````

````python
print(np.log(np.mean(data[: , 0:1])) ) # Example of calculating the natural logarithm of the mean of the first column
````
