```python
# elif
exp = 16

if exp >= 0 and exp < 1:
    print("Fresher & eligible for internship")
elif exp >= 1 and exp < 3:
    print("Junior & eligible for data Analyst role")
elif exp >= 3 and exp < 5:
    print("Mid-level & eligible for data Scientist role")
elif exp >= 5 and exp < 10:
    print("Senior & eligible for AI Engineer role")
elif exp >= 10:
    print("Expert & eligible for Manager role")
else:
    print("Invalid experience level")
```

```python
card = 'access'
atm_pin = 1234
balance = 30000
withdrawal_amount = 5000
```

```python
# Nested if-else
if card == 'access':
    if atm_pin == 1234:
        if balance >= withdrawal_amount:
            balance -= withdrawal_amount
            print(f"Withdrawal successful! New balance: {balance}")
        else:
            print(f"Insufficient balance! Current balance: {balance}")
    else:
        print("Incorrect ATM pin")
else:
    print("Card not recognized")
```

### How to print values in the same line in a for loop
Use the `end` parameter in the `print()` function, e.g., `print(i, end=' ')`.

```python
for i in age:
    print(i, end=' ')
```

```python
for i in age:
    if i > (sum(age) / len(age)): # Compare with average, sum(age) gives total, len(age) gives count
        print(i, end=' ')
```
