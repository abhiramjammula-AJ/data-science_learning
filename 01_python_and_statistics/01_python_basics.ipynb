```python
# Calculate sum and mean of a list
numbers = [10, 20, 30, 40, 50]
total = sum(numbers)
mean = total / len(numbers)
print("Sum:", total)
print("Mean:", mean)
```

```python
# String operations example
str1 = "Hello"
str2 = "World"
combined = str1 + " " + str2
print("Combined:", combined)
print("Length:", len(combined))
print("Uppercase:", combined.upper())

# Boolean operations example
a = True
b = False
print("a AND b:", a and b)
print("a OR b:", a or b)
print("NOT a:", not a)
```

```python
# Arithmetic operations
a = 7
b = 5
print("Sum:", a + b)
print("Difference:", a - b)
print("Product:", a * b)
print("Quotient:", a / b)
print("Modulus:", a % b)
print("Power:", a ** b)
print("Floor division:", a // b)
```

```python
# Comparison operations
a = 10
b = 5
print("a > b:", a > b)
print("a < b:", a < b)
print("a == b:", a == b)
```

```python
# Logical operations
a = True
b = False
print("a AND b:", a and b)
print("a OR b:", a or b)
print("NOT a:", not a)
```

```python
# Bitwise operations
a = 7
b = 5
print("a & b:", a & b) # Bitwise AND
print("a | b:", a | b) # Bitwise OR
print("a ^ b:", a ^ b) # Bitwise XOR
```

```python
# Membership operations
my_list = [1, 2, 3, 4, 5]
print("2 in my_list:", 2 in my_list)
print("6 not in my_list:", 6 not in my_list)
```

```python
# Identity operations
a = 10
b = 5
print("a is b:", a is b)
print("a is not b:", a is not b)
```

```python
# Assignment operations
x = 10
y = 20
x += 5  # x = x + 5
y -= 10  # y = y - 10
print("x after += 5:", x)
print("y after -= 10:", y)
```
