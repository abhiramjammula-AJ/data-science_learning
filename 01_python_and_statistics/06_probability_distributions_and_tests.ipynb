### Normal Distribution

```python
import pandas as pd
import numpy as np
from scipy.stats import norm
from statsmodels.stats.weightstats import ztest
```

```python
# norm(mean, std_dev) creates a normal distribution object
# mean = 50000, std_dev = 10000
nd = norm(50000, 10000) 
```

cdf(50000) adds the probability of all values less than 50000  
nd.cdf() # cumulative distribution function

```python
#  P(X > 50000)
nd.cdf(50000) #cumelative distribution function
```

we dont heve function for greater than so 1-nd.cdf()

```python
# P(X > 50000)
# 1-cdf(50000) adds the probability of all values greater than than 50000
1-nd.cdf(50000)
```

```python
# P(x=50000)
print("Nearly 0")
nd.pdf(50000) # probability density function
```

Hypothesis Testing  
  
# one sample Z test

```python
df = pd.read_csv("D:\\Data science\\codes\\files\\Bulbs_Lifespan.csv")
df
```

```python
z_cal, pvalue = ztest(df["Lifespan (Hours)"],value=980,alternative="two-sided")
```

```python
print("z cal:", np.round(z_cal,2))
```

```python
print("pvalue:", np.round(pvalue,2))
```

```python
alpha = 0.05
if pvalue < alpha:
    print("Ho is rejected and H1 is accepted")
else:
    print("H1 is rejected and Ho is accepted")
```

# Two sample z-test

```python
df = pd.read_csv("D:\\Data science\\codes\\files\\ecommerce_two_sample_ztest_dataset.csv")
df
```

```python
df["Group"] == "Email Marketing"
```

```python
d1 = df[df["Group"] == "Email Marketing"]
d2 = df[df["Group"]=="Social Media Ads"]
```

```python
d1["Purchase_Value"]
```

```python
d2["Purchase_Value"]
```

```python
z_stat, pvalue = ztest(d1["Purchase_Value"], d2["Purchase_Value"], alternative="two-sided")
```

```python
print(f"z stat: {z_stat}")
print(f"pvalue: {pvalue}")
```

```python
if pvalue < 0.05:
    print("H0 is Rejected")
else:
    print("H0 is accepted")
```
