### Importing into python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Cardiotocographic.csv")
```

```python
df
```

### Handling the missing values

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
m1 = df['LB'].mean()
m2 = df['AC'].mean()
m3 = df['DS'].mean()
m4 = df['DP'].mean()
m5 = df['MLTV'].mean()
m6 = df['Width'].mean()
m7 = df['Tendency'].mean()
m8 = df['NSP'].mean()
```

```python
df['LB'].fillna(m1,inplace=True)
df['AC'].fillna(m2,inplace=True)
df['DS'].fillna(m3,inplace=True)
df['DP'].fillna(m4,inplace=True)
df['MLTV'].fillna(m5,inplace=True)
df['Width'].fillna(m6,inplace=True)
df['Tendency'].fillna(m7,inplace=True)
df['NSP'].fillna(m8,inplace=True)
```

```python
df.isnull().sum()
```

### statistical summary for each variable

```python
df.describe()
```

### data visualization

Histogram

```python
df.hist(figsize=(15,10))
plt.tight_layout()
plt.show()
```

BoxPlot

```python
df.plot(kind='box', subplots=True, layout=(4, 4), figsize=(15, 10), vert = False, sharex=False, sharey=False)
plt.tight_layout()
plt.show()
```

HeatMap using seaborn

```python
plt.figure(figsize=(12, 8))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Variables')
plt.show()
```
