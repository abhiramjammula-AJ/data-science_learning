### Exercise 1: Prime number

```python
def isPrime(num):
    # Function to check if a number is prime or not

    if num > 3:
        # If number is greater than 3, we check divisibility
        for i in range(2, num - 1):  
            # Loop from 2 to num-1 to check if any number divides num
            if (num % i == 0):
                # If remainder is 0, num is divisible by i (not prime)
                print(num, "is not a prime number")
                break  # No need to check further, exit loop
            else:
                # If not divisible by the current i, assume prime here
                print(num, "is a prime number")
                break
    elif num < 1:
        # If the number is less than 1, it's invalid for prime checking
        print("Enter the number greater than or equal to 1")
    else:
        # For numbers 1, 2, or 3 — considered prime by this code's logic
        print(f"{num} is a prime number")


# Taking user input as an integer
n = int(input())

# Calling the isPrime function with user input
isPrime(n)
```

### Exercise 2: Product of Random Numbers

```python
import numpy as np  # Importing NumPy library for random number generation

def randomNumProduct():
    # Function to generate two random integers between 1 and 5 (inclusive)
    
    n1 = np.random.randint(1, 6, 1)  # Random integer between 1 and 5 for first number
    n2 = np.random.randint(1, 6, 1)  # Random integer between 1 and 5 for second number
    
    product = n1 * n2  # Calculate the product of the two random numbers
    
    # Display the generated numbers
    print(f"product number are {n1} & {n2}")
    
    return product  # Return the product of the two numbers


# Ask the user to guess the product of two random numbers between 1 and 5
n = int(input("Enter the guessed answer of product between two numbers that are between 1 to 5"))

# Generate random numbers and get their product
product = randomNumProduct()

# Compare user's guess with the actual product
if (n == product):
    # If guess is correct
    print("Congradulations you have predected correctly")
    print("The answer is", product)
    print("And you guessed", n)
else:
    # If guess is incorrect
    print("awww looks like you missed it this time")
    print("The correct answer was", product)
    print("you guessed", n)
```

### Exercise 3: Squares of Even/Odd Numbers

```python
def squareOdd():
    # Function to print squares of all odd numbers between 100 and 200
    print("printing squares of all odd numbers between 100 to 200")
    for i in range(101, 201, 2):  # Start from 101, step by 2 to get odd numbers
        print(f"square of {i} = {i*i}")  # Calculate and print square


def squareEven():
    # Function to print squares of all even numbers between 100 and 200
    print("Printing squares of all even numbers between 100 to 200")
    for i in range(100, 201, 2):  # Start from 100, step by 2 to get even numbers
        print(f"square of {i} = {i*i}")  # Calculate and print square


# Display menu for the user
print("1) To get all the squares of even number between 100 to 200 ")
print("2) To get all the squares of odd number between 100 to 200 ")

# Take user input for choice
n = int(input())

# Based on user's choice, call the corresponding function
if n == 1:
    squareEven()
elif n == 2:
    squareOdd()
else:
    # If input is not 1 or 2, show invalid choice
    print("Invalid choice")
```

### Exercise 4: Word counter

```python
# Take text input from the user
input_text = input("Enter a sentence or paragraph: ")

# Split the text into words using spaces as separators
words = input_text.split()

# Create an empty dictionary to store word counts
word_count = {}

# Loop through each word in the list
for word in words:
    # If the word is already in the dictionary, increase its count
    if word in word_count:
        word_count[word] += 1
    else:
        # If the word is new, set its count to 1
        word_count[word] = 1

# Print each word and its count
print("\nWord Count Result:")
for word, count in word_count.items():
    print(f"'{word}': {count}")
```

### Exercise 5: Check for Palindrome


```python
# Take user input
word = input("Enter a word")

# Remove spaces and convert to lowercase for accurate comparison of a phrase just for additonal
# processed_word = word.replace(" ", "").lower() 
word = word.lower()
# Reverse the processed string
reversed_word = word[::-1]

# Check if the original processed word matches the reversed one
if word == reversed_word:
    print(f"'{word}' is a palindrome")
else:
    print(f"'{word}' is not a palindrome")
```
