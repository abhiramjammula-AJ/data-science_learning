##### 6.	Given a 'marks' variable, print the grade:
- 90 and above: 'A'
- 75–89: 'B'
- 50–74: 'C'
- Below 50: 'Fail'


```python
grade = int(input())
if 90 <= grade <= 100:
    print("Grade: A")
elif 75<= grade <=89:
    print("Grade: B")
elif 50<= grade <= 74:
    print("Grade: C")
elif 0<= grade <50:
    print("Grade: Fail")
else:
    print("marks must be between range(0,100)")
```

##### 7.	Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
num = int(input())
if num > 0:
    if num%2 == 0:
        print(f"{num} is both positive and even")
    else:
        print(f"{num} is positive, but odd")
elif num == 0:
    print(f"number is {num}")
else:
    print(f"{num} is nevative")
```
