````python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Sample sales data
sales = [200, 300, 150, 400, 250]

# Descriptive Analytics: Calculate mean
average_sales = np.mean(sales)
print(f"Average daily sales: ${average_sales}")

# Visualize sales with a bar chart
days = ['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5']
plt.bar(days, sales, color='skyblue')
plt.title('Daily Sales')
plt.xlabel('Day')
plt.ylabel('Sales ($)')
plt.show()
````

## Line-by-Line Code Explanation

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt

# Sample sales data
sales = [200, 300, 150, 400, 250]

# Descriptive Analytics: Calculate mean
average_sales = np.mean(sales)
print(f"Average daily sales: ${average_sales}")

# Visualize sales with a bar chart
days = ['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5']
plt.bar(days, sales, color='skyblue')
plt.title('Daily Sales')
plt.xlabel('Day')
plt.ylabel('Sales ($)')
plt.show()
```

- `import numpy as np`: Imports the numpy library and gives it the alias `np` for numerical operations.
- `import matplotlib.pyplot as plt`: Imports the plotting library for creating charts.
- `sales = [200, 300, 150, 400, 250]`: Creates a list of sales numbers for five days.
- `average_sales = np.mean(sales)`: Calculates the average of the sales list using numpy's mean function.
- `print(f"Average daily sales: ${average_sales}")`: Prints the calculated average sales.
- `days = ['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5']`: Creates a list of day labels for the x-axis.
- `plt.bar(days, sales, color='skyblue')`: Draws a bar chart with days on the x-axis and sales on the y-axis.
- `plt.title('Daily Sales')`: Sets the title of the chart.
- `plt.xlabel('Day')`: Labels the x-axis as 'Day'.
- `plt.ylabel('Sales ($)')`: Labels the y-axis as 'Sales ($)'.
- `plt.show()`: Displays the chart.

````python
import numpy as np
import matplotlib.pyplot as plt

sales = [240, 170, 290, 320, 410, 190, 200]
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
average_sales = np.mean(sales)
print(f"Average sales in a week: ${average_sales}")

plt.bar(days, sales, color='green')
plt.title("Weekly Sales")
plt.xlabel('Day')
plt.ylabel('Sales (Rs)')
plt.show()
````

````python
# Measures of Central Tendency for Volume
mean_volume = df['Volume'].mean()
median_volume = df['Volume'].median()
mode_volume = stats.mode(df['Volume'])  # Mode returns an array, take first value

print(f"Mean Volume: {mean_volume}")
print(f"Median Volume: {median_volume}")
print(f"Mode Volume: {mode_volume}")
````

````python
range = df['Volume'].max() - df['Volume'].min()
variance = df['Volume'].var()
std = df['Volume'].std()
print(f"Variance = {variance}")
print(f"Standard Deviation = {std}")
print(f"Range of volume: {range}")
````
