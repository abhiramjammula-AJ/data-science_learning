```python
import nltk
nltk.download('all')
```

```python
text = "I got a great deal on this Laptop, which I have been using for almost a year! now. It is been working well but needs some customization"
```

### Tokenization

```python
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
words = word_tokenize(text)
display(words)
words0 = sent_tokenize(text)
display(words0)
```

```python
display(len(words))
# sencance tokenization is also available
```

### Normalization

transform various forms of same term into a common 'Normalized form

```python
# either upper to lower / lower to upper
words2 = [X.upper() for X in words]
print(words2)
words2 = [X.lower() for X in words]
print(words2)
```

```python
import string

words3 = [X for X in words2 if X not in string.punctuation]
display(len(words3))
words3
```

```python
display(string.punctuation)
print("after removing punctuation:",len(words3))
```

### stop words

list of english words which helping us to make a proper sentance  

always convert the word to lower cases or convert stopwords to upper cases

```python
from nltk.corpus import stopwords
stopwords_list = stopwords.words('english')
display(len(stopwords_list))
stopwords_list
# since it is a list we also can add some words if we need using .append
```

```python
words4 = [X for X in words3 if X not in stopwords_list]
display(words4)
```

```python
len(words4)
```

### Lemmatization and stemming

same meaning but differt forms like walk : walking : walked  
so stemming chops of the right end to remove lingustic forms  

some word may not required cutting so lemmatization checks if chopping is required, if not it remains the words as it is  

eg: i am watching at a watch  
stemming makes it as: i am watch at a watch (reduces the toneization)  
lemmatization makes it as: i am watching the watch --> as it is

```python
from collections import Counter
words_counts = Counter(words4)
display(words_counts)
```

```python
e_words = ["wait","waiting","waited", "waits", "waiter"]
from nltk.stem import PorterStemmer
p_stemmer = PorterStemmer()

for X in e_words:
    print(X+' __> '+p_stemmer.stem(X))
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
for X in e_words:
    print(f"{X:{10}} --> {lemmatizer.lemmatize(X)}")
```
