# Visulization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib import pyplot
```

```python
excel_file = pd.ExcelFile("D:\\Data science\\codes\\files\\daily-minimum-temperatures.xls")
excel_file.sheet_names
```

```python
df = excel_file.parse('daily-minimum-temperatures')
df = df.drop(columns=['Unnamed: 2', 'lag 1', 'lag2', 'lag 3'])
df.to_csv("D:\\Data science\\codes\\files\\daily-minimum-temperatures.csv")
```

```python
df.head()
```

```python
df.plot(y='Temp')
```

```python
df.hist()
pyplot.show()
```

```python
df['Temp'].skew()
```

```python
df.describe()
```

```python
df.plot(y='Temp',kind='kde')
pyplot.show()
```

```python
df['year'] = pd.to_datetime(df['Date']).dt.year
df
```

```python
plt.figure(figsize=(12,8))
sns.boxplot(x='year',y='Temp',data=df)
plt.xlabel('Year')
plt.ylabel('Temperature')
plt.title("year-wise Box plot of temp")
plt.show()
```

```python
df['month'] = pd.to_datetime(df['Date']).dt.month
df
```

```python
plt.figure(figsize=(12,8))
sns.boxplot(x='month',y='Temp',data=df)
plt.xlabel('month')
plt.ylabel('Temperature')
plt.title("month-wise Box plot of temp")
plt.show()
```

# Lag plotting

```python
from pandas.plotting import lag_plot
lag_plot(df['Temp'])
plt.show()
```
