```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

### Label Encoder

```python
df = pd.read_csv("D:\\Data science\\codes\\files\\Diab.csv")
df
```

```python
LE = LabelEncoder()
LE.fit_transform(df["Gender"])
```

```python
df["Gender_LE"] = LE.fit_transform(df["Gender"])
df
```

### One Hot Encoder

```python
OHE = OneHotEncoder()
OHE.fit_transform(df[["Gender"]])
```

```python
OHE.fit_transform(df[["Gender"]]).toarray() # to make it an array
```

```python
d1 = OHE.fit_transform(df[["Gender"]]).toarray()
d2 = OHE.fit_transform(df[["Diabetic"]]).toarray()
d1 = pd.DataFrame(d1)
d2 = pd.DataFrame(d2)
# we cant directly push into the data since it is an array
# so we need to convert it into a dataframe first
# we get as per alphabetical
d1.columns = ["G_Female", "G_Male"]
d2.columns = ["D_no", "D_Yes"]
df_new = pd.concat([df,d1,d2], axis=1)
df_new
```
