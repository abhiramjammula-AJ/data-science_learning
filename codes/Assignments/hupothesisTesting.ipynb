Hypothesis Testing for Bombay Hospitality Ltd.
This script performs hypothesis testing to investigate whether the weekly operating costs
are higher than the theoretical model W = $1,000 + $5X.

we perform one sample Z test

Step 1: State the Hypotheses
Null Hypothesis (H0): The mean weekly operating cost is less than or equals the theoretical cost (μ <= $1,000 + $5 * 600 <= $4,000)
Alternative Hypothesis (H1): The mean weekly operating cost is greater than the theoretical cost (μ > $4,000)
This is a one-tailed test (right-tailed) since we're testing if costs are higher. 


```python
# given data
sample_mean = 3050  # Sample mean weekly cost (Rs. 3,050)
theoretical_x = 600  # Mean number of units produced (X)
std_dev_x = 25  # Standard deviation of units produced
sample_size = 25  # Number of restaurants
alpha = 0.05  # Significance level
```

Step 2: Calculate the Test Statistic  
Theoretical mean weekly cost (μ) based on W = $1,000 + $5X  
For X = 600 units, μ = 1000 + 5 * 600 = 1000 + 3000 = $4,000

```python
theoretical_mean = 1000 + 5 * theoretical_x
# Standard deviation of weekly cost (σ) = 5 * std_dev_X = 5 * 25 = 125
std_dev_cost = 5 * std_dev_x
# Test statistic formula: t = (sample_mean - theoretical_mean) / (std_dev_cost / sqrt(sample_size))
standard_error = std_dev_cost / np.sqrt(sample_size)
test_statistic = (sample_mean - theoretical_mean) / standard_error
```

Step 3: Determine the Critical Value  
For a one-tailed test with α = 0.05, we use the Z-distribution (since sample size is given and population standard deviation is known)  
Critical value for a right-tailed test at α = 0.05 is approximately 1.65 (from standard normal Z-table)

```python
critical_value = 1.65
```

Step 4: Make a Decision  
If test_statistic > critical_value, reject the null hypothesis  
Otherwise, fail to reject the null hypothesis

```python
decision = "Reject the null hypothesis" if test_statistic > critical_value else "Fail to reject the null hypothesis"
```

Step 5: Conclusion  
If we reject H0, there is strong evidence that weekly operating costs are higher than the model suggests 


```python
conclusion = (
    "There is strong evidence to support the restaurant owners' claim that the weekly operating costs are higher than the model suggests." if test_statistic > critical_value else "There is not enough evidence to support the restaurant owners' claim that the weekly operating costs are higher than the model suggests."
)
```

Print results

```python
# Print results
print("Hypothesis Testing Results using ztest")
print("=====================================")
print(f"Step 1: Hypotheses")
print(f"Null Hypothesis (H0): Mean weekly cost <= ${theoretical_mean}")
print(f"Alternative Hypothesis (H1): Mean weekly cost > {theoretical_mean}")
print(f"\nStep 2: Test Statistic")
print(f"Theoretical Mean (μ): ${theoretical_mean}")
print(f"Standard Deviation of Cost (σ): ${std_dev_cost}")
print(f"Test Statistic (z): {test_statistic}")
print(f"\nStep 3: Critical Value")
print(f"Critical Value (Z, α = 0.05, one-tailed): {critical_value}")
print(f"\nStep 4: Decision")
print(f"Decision: {decision}")
print(f"\nStep 5: Conclusion")
print(f"Conclusion: {conclusion}")
print(f"\nThe test statistic ({test_statistic:.2f}) is compared to the critical value ({critical_value}). "
      f"\nSince {test_statistic:.2f} < {critical_value}, we fail to reject H0, indicating insufficient evidence "
      f"to conclude that costs are higher than the model.")
```
