```python
numerical_col = df.columns[[4,8,9,10,11,12]]
```

```python
for col in numerical_col:
    plt.figure()
    df[col].hist()
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for col in numerical_col:
    plt.figure()
    df.boxplot(column=col, vert = False)
    plt.title(f"Boxplot of {col}")
    plt.show()
```
