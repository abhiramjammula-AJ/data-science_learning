```python
import numpy as np
from scipy import stats
```

# Data

```python
durability_data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
```

# a. Build 99% Confidence Interval Using Sample Standard Deviation


```python
# Calculate sample mean and sample standard deviation
sample_mean = np.mean(durability_data)
sample_std = np.std(durability_data, ddof=1) # Use ddof=1 for sample standard deviation
n = len(durability_data)
confidence_level = 0.99
alpha = 1 - confidence_level
degrees_of_freedom = n - 1
```

```python
# Find the critical t-value
t_critical = stats.t.ppf(1 - alpha/2, degrees_of_freedom)

# Calculate the margin of error
margin_of_error_t = t_critical * (sample_std / np.sqrt(n))

# Calculate the confidence interval
confidence_interval_t = (sample_mean - margin_of_error_t, sample_mean + margin_of_error_t)

print("a. 99% Confidence Interval Using Sample Standard Deviation:")
print(f"Sample Mean: {sample_mean:.4f}")
print(f"Sample Standard Deviation: {sample_std:.4f}")
print(f"Degrees of Freedom: {degrees_of_freedom}")
print(f"Critical t-value: {t_critical:.4f}")
print(f"Margin of Error (t-distribution): {margin_of_error_t:.4f}")
print(f"Confidence Interval (t-distribution): ({confidence_interval_t[0]:.4f}, {confidence_interval_t[1]:.4f})")
print("-" * 50)
```

# b. Build 99% Confidence Interval Using Known Population Standard Deviation


```python
# Known population standard deviation
population_std = 0.2

# Find the critical z-value
z_critical = stats.norm.ppf(1 - alpha/2)

# Calculate the margin of error
margin_of_error_z = z_critical * (population_std / np.sqrt(n))

# Calculate the confidence interval
confidence_interval_z = (sample_mean - margin_of_error_z, sample_mean + margin_of_error_z)

print("b. 99% Confidence Interval Using Known Population Standard Deviation:")
print(f"Sample Mean: {sample_mean:.4f}")
print(f"Known Population Standard Deviation: {population_std}")
print(f"Critical z-value: {z_critical:.4f}")
print(f"Margin of Error (z-distribution): {margin_of_error_z:.4f}")
print(f"Confidence Interval (z-distribution): ({confidence_interval_z[0]:.4f}, {confidence_interval_z[1]:.4f})")
```
